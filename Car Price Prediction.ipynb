```python
x = df[['make', 'fuel-type', 'aspiration', 'num-of-doors', 'body-style',
       'drive-wheels', 'engine-location', 'wheel-base', 'length', 'width',
       'curb-weight', 'engine-type', 'num-of-cylinders',
       'engine-size', 'fuel-system', 'bore',
       'horsepower','city-mpg', 'highway-mpg']]

y = df[['price']]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder,LabelEncoder,OrdinalEncoder
from sklearn.compose import ColumnTransformer
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,train_size = 0.70,random_state = 2)
```

```python
x_train
```

```python
df_doors = df[['num-of-doors']]
```

```python
OE__doors = OrdinalEncoder(categories = [['two','four']])
```

```python
df_doors["num-of-doors"] = OE__doors.fit_transform(df_doors[['num-of-doors']])
```

```python
x_train['num-of-doors'] = OE__doors.fit_transform(x_train[['num-of-doors']])
```

```python
x_test['num-of-doors'] = OE__doors.transform(x_test[['num-of-doors']])
```

```python
x_train['num-of-cylinders'].value_counts()
```

```python
OE_cylinder = OrdinalEncoder(categories = [['three',"four",'five','six','eight']])
```

```python
x_train["num-of-cylinders"] = OE_cylinder.fit_transform(x_train[["num-of-cylinders"]])
```

```python
x_test["num-of-cylinders"] = OE_cylinder.transform(x_test[["num-of-cylinders"]])
```

```python
x_train.select_dtypes(include = 'object').columns.to_list()
```

```python
transformer = ColumnTransformer(transformers = [
    ('tfn1',OneHotEncoder(sparse = False,drop = 'first'),['make','fuel-type','aspiration','body-style','drive-wheels',
                                                          'engine-location','engine-type','fuel-system'])],remainder = 'passthrough')
```

```python
X_train_transformed = transformer.fit_transform(x_train)
```

```python
X_test_transformed = transformer.transform(x_test)
```

```python
X_train_transformed.shape
```

```python
X_train_transformed
```

```python
X_test_transformed
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train_transformed,y_train)
```

```python
y_pred = lr.predict(X_test_transformed)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
y_train_pred = lr.predict(X_train_transformed)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
print("======Performance Card======")
print("Mean Absolute Error : ",mean_absolute_error(y_test,y_pred))
print("Mean Squared Error : ",mean_squared_error(y_test,y_pred))
print("Root Mean Squared Error : ",np.sqrt(mean_squared_error(y_test,y_pred)))
print("R2 Score : ",r2_score(y_test,y_pred))
```

```python
plt.figure(figsize=(6, 6))
plt.scatter(y_test, y_pred, color='blue', alpha=0.6)
plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Actual vs Predicted Values (Multiple Linear Regression)")

# 45-degree reference line (perfect predictions)
plt.plot([y_test.min(), y_test.max()],
         [y_test.min(), y_test.max()],
         color='red', linestyle='--')

plt.show()
```
